# Latest PySpark Test Run Dashboard

Run `run_test_notebook.ipynb` first. This dashboard reads the latest saved pytest result from Delta tables.

```python
from pyspark.sql import functions as F

SUMMARY_TABLE = "pyspark_demo_test_run_results"
DETAILS_TABLE = "pyspark_demo_test_case_results"

summary = spark.table(SUMMARY_TABLE)
latest_run = summary.orderBy(F.col("finished_at").desc()).limit(1)

display(latest_run)
```

```python
latest_run_id = latest_run.select("run_id").first()["run_id"]

details = (
    spark.table(DETAILS_TABLE)
    .where(F.col("run_id") == latest_run_id)
    .orderBy(F.col("outcome"), F.col("duration_seconds").desc())
)

display(details)
```
